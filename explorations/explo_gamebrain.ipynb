```python
import os
import pandas as pd
from dotenv import load_dotenv
import requests
import json

load_dotenv()
API_KEY = os.getenv("GAMEBRAIN_API_KEY")

try:
    print(f"GAMEBRAIN_API_KEY: {API_KEY}")
    if API_KEY is None:
        raise ValueError("GAMESDB_API_KEY is not set in the environment variables.")
except Exception as e:
    print(f"Error occurred: {e}")
```

```python
url = "https://api.gamebrain.co/v1/games?query=rayman origins"

headers = {
    "x-api-key": API_KEY
}

response = requests.get(url, headers=headers)
if response.status_code == 200:
    data = response.json()
    print(data)
    games = data.get("results", [])
    df_brain = pd.DataFrame(games)
    print(df_brain.columns.tolist())
else:
    print(f"Error: {response.status_code} - {response.text}")
```
