```python
import os
import pandas as pd
from dotenv import load_dotenv
import requests
import json

load_dotenv()
API_KEY = os.getenv("GAMESDB_API_KEY")

try:
    print(f"GAMESDB_API_KEY: {API_KEY}")
    if API_KEY is None:
        raise ValueError("GAMESDB_API_KEY is not set in the environment variables.")
except Exception as e:
    print(f"Error occurred: {e}")
```

```python
url = "https://api.thegamesdb.net/v1.1/Games/ByGameName"
response = requests.get(url, params={"apikey": API_KEY,"name": "the witcher", "filter[region_id]":"6","fields":"publishers,genres,overview,platform,coop,release_date,developers,tags"})
if response.status_code == 200:
    data = response.json()
    print(data)
    games = data.get("data", {"games":[]}).get("games", [])
    df_gamesdb = pd.DataFrame(games)
    print(df_gamesdb.columns.tolist())
else:
    print(f"Error: {response.status_code} - {response.text}")
```

```python
url = "https://api.thegamesdb.net/v1/Games/ByGameID"
response = requests.get(url, params={"apikey": API_KEY,"id": "5754", "fields": "publishers, genres, developers, overview", "include": "boxart, platforms"})
if response.status_code == 200:
    data = response.json()
    print(data)
    game = data.get("data", {"games":[]}).get("games", [])
    df_game = pd.DataFrame(game)
    print(df_game.columns.tolist())
else:
    print(f"Error: {response.status_code} - {response.text}")
```
